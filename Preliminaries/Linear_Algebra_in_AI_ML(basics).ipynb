##Basic Properties of tensor arithmetic

```python
A = torch.arange(6, dtype=torch.float32).reshape(2, 3)
B = A.clone() # Assign a copy of A to B by allocating new memory
A, A + B
```

```python
A * B  # Hadamard product = elementwise product of two matrices
```

```python
a = 2
X = torch.arange(24).reshape(2, 3, 4)
a + X, (a * X).shape, len(X)
```

##Reduction

```python
x = torch.arange(3, dtype=torch.float32)
x, x.sum()
```

```python
A.shape, A.sum()     # sums over the elements of tensors of arbitrary shape
```

```python
A.shape, A.sum(axis=0).shape  # specify the axes along which the tensor should be reduced
```

```python
A.shape, A.sum(axis=1).shape
```

```python
A.sum(axis=[0, 1]) == A.sum()   # Same as A.sum()
```

```python
A.mean(), A.sum() / A.numel()
```

```python
A.mean(axis=0), A.sum(axis=0) / A.shape[0]
```

##Non-Reduction Sum

```python
sum_A = A.sum(axis=1, keepdims=True)   # keep the number of axes unchanged
sum_A, sum_A.shape
```

```python
A / sum_A
```

```python
  A.cumsum(axis=0)  # the cumulative some of elements of A along some axis
```

##Dot Products

```python
y = torch.ones(3, dtype=torch.float32) # The dot product of two vectors is a sum over the products of the elements at the same position
x, y, torch.dot(x, y)
```

```python
torch.sum(x * y)  # equivalent way of finding dot product
```

## Matrix -- Vector products

```python
A.shape, x.shape, torch.mv(A, x), A@x      # m*n matrix and n-dimensional vector product
```

## Matrix -- Matrix multiplication

```python
B = torch.ones(3, 4)          # A(m*k), B(k*n), so basically while multiplying both, we multiply row vectors of A and column vectors of B
torch.mm(A, B), A@B
```

##Norms

```python
u = torch.tensor([3.0, -4.0])      # l2 norm(Euclidean norm) where we get root of the sum of squared elements
torch.norm(u)
```

```python
torch.abs(u).sum()     # l1 norm(Manhattan distance) where it sums the absolute values of vector's elements
```

```python
torch.norm(torch.ones((4, 9)))    # For Matrices Frobenius norm is used to calculate root of the sum of the all elemnts of the matrix
```
