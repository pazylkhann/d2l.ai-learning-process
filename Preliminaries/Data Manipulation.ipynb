##Saving Memory

```python
before = id(Y)
Y = Y + X
id(Y) == before           # Proving that after we get new allocated memory and and then point variable to new allocated memory
```

```python
Z = torch.zeros_like(Y)        # performing in-place operation
print("id(Z):", id(Z))
Z[:] = X + Y
print("id(z)", id(Z))
# We can assign the result of an operation to a previously allocated array Y by using slice notation: Y[:] = <expression>.
# To illustrate this concept, we overwrite the values of tensor Z, after initializing it, using zeros_like, to have the same shape as Y.
```

```python
before = id(X)
X += Y       # reducing memory overhead of the operation if there is no subsequent computations
id(X) == before
```

## Conversion to other Python Objects

```python
A = X.numpy()
B = torch.from_numpy(A)
type(A), type(B)
#  The torch tensor and NumPy array will share their underlying memory, and
# Changing one through an in-place operation will also change the other.
```
